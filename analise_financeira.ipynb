```python
import pandas as pd
```

```python
dados = {
    'Nome': ['Lucas Silva', 'Mariana Costa', 'Beatriz Rocha', 'Rafael Alves', 'Camila Lima', 'Thiago Santos', 'Juliana Melo'],
    'Salario_Mensal': [4500, 3200, 6000, 2800, 5500, 3800, 7200],
    'Jan': [3100, 3400, 4200, 2900, 4000, 3700, 5000],
    'Fev': [2900, 3500, 5100, 2850, 3900, 4100, 5200],
    'Mar': [3200, 3100, 4800, 3000, 4100, 3900, 4900],
    'Abr': [3000, 3300, 4000, 2950, 3800, 4000, 5100],
    'Mai': [2800, 3600, 4500, 3100, 4200, 3800, 5300],
    'Jun': [3100, 3200, 4100, 2800, 3900, 4200, 4800]
}
```

```python
df = pd.DataFrame(dados)
```

```python
df
```

```python
df["Renda_Semestral"] = df["Salario_Mensal"] *6
```

```python
df
```

```python
df["Gastos_Totais"] = df[["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"]].sum(axis=1)
```

```python
df["Saldo_Final"] = df ["Renda_Semestral"] - df["Gastos_Totais"]
```

```python
df
```

```python
df["Status"] = df["Saldo_Final"].apply(lambda x: "Bom Semestre" if x >= 0 else "Com Divídas")
```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

df_sorted = df.sort_values(by='Saldo_Final', ascending=True)

cores = ['#2ecc71' if saldo >= 0 else '#e74c3c' for saldo in df_sorted['Saldo_Final']]

plt.figure(figsize=(11, 6))

barras = plt.barh(df_sorted['Nome'], df_sorted['Saldo_Final'], color=cores)

plt.axvline(0, color='black', linewidth=1.2, linestyle='--')

for barra in barras:
    largura = barra.get_width()

    if largura >= 0:
        posicao_x = largura + 300
        ha_alinhamento = 'left'
    else:
        posicao_x = largura - 300
        ha_alinhamento = 'right'

    plt.text(
        posicao_x,
        barra.get_y() + barra.get_height() / 2,
        f'R$ {largura:,.0f}',
        va='center',
        ha=ha_alinhamento,
        fontsize=10,
        fontweight='bold'
    )

menor_valor = df_sorted['Saldo_Final'].min()
maior_valor = df_sorted['Saldo_Final'].max()
plt.xlim(menor_valor - 2500, maior_valor + 2500)

plt.title('Balanço Final do Semestre (Saldo em R$)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Saldo Final (Verde = Sobrou | Vermelho = Dívida)', fontsize=11)
plt.ylabel('Pessoa', fontsize=11)
plt.grid(axis='x', linestyle=':', alpha=0.6)
plt.tight_layout()

plt.show()
```
